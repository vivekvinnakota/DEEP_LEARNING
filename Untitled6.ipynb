```python
import torch
import pandas as pd
import plotly.express as px
import matplotlib.pyplot as plt

from PIL import Image
from transformers import CLIPProcessor, CLIPModel
from google.colab import files
from IPython.display import display

device = "cuda" if torch.cuda.is_available() else "cpu"

model_name = "openai/clip-vit-base-patch32"

model = CLIPModel.from_pretrained(model_name)
processor = CLIPProcessor.from_pretrained(model_name)

model.to(device)
model.eval()

print("Model loaded successfully!")
print("Device:", device)
```

```python
uploaded = files.upload()

image_names = list(uploaded.keys())

image1 = Image.open(image_names[0]).convert("RGB")
image2 = Image.open(image_names[1]).convert("RGB")
```

```python
animals = [
    "dog", "cat", "lion", "tiger",
    "elephant", "horse", "cow", "deer",
    "zebra", "giraffe", "monkey"
]

labels = [
    "car",
    *animals,
    "other object"
]

descriptions = [
    "a photograph of a car",
    *[f"a photograph of a {animal}" for animal in animals],
    "a photograph of an object that is neither an animal nor a car"
]
```

```python
def classify_image(image):

    inputs = processor(
        text=descriptions,
        images=image,
        return_tensors="pt",
        padding=True
    ).to(device)

    with torch.no_grad():

        outputs = model(**inputs)

        probabilities = outputs.logits_per_image.softmax(dim=1)[0]

    scores = probabilities.cpu().numpy()

    results = pd.DataFrame({
        "Class": labels,
        "Confidence": scores * 100
    })

    results = results.sort_values(
        by="Confidence",
        ascending=False
    )

    prediction = results.iloc[0]["Class"]

    return prediction, results
```

```python
prediction1, results1 = classify_image(image1)
prediction2, results2 = classify_image(image2)

if prediction1 == prediction2:
    print("Both images belong to the same predicted class!")
else:
    print("The images belong to different predicted classes!")

print("Image 1:", prediction1)
print("Image 2:", prediction2)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

for ax, results, title in [
    (axes[0], results1, "Image 1 Predictions"),
    (axes[1], results2, "Image 2 Predictions")
]:

    top = results.head(8).sort_values("Confidence")

    ax.barh(top["Class"], top["Confidence"])

    ax.set_title(title)
    ax.set_xlabel("Relative score (%)")

plt.tight_layout()
plt.show()
```
